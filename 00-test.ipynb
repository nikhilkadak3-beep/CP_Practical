```python
# get marks of student and give grades based on marks 
# above 95 -grade A , above 85 - grade B , above 75 - grade c , above 55 - grade D 
marks = int(input("Enter the marks of the student: "))
if marks >= 95:
    print("Grade A")
elif marks >= 85:
    print("Grade B")
elif marks >= 75:
    print("Grade C")
elif marks >= 65:
    print("Grade D")
elif marks >= 55:
    print("Grade E")        
```
